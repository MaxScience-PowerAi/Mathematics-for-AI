## 3 / Quadratic optimization
The update factor is $1-\eta\kappa$. Observe convergence and divergence for a nonzero initial point.

```python
fig, ax = plt.subplots(figsize=(8,4))
for eta in [.25,.5,.75,1.,1.2]:
    x, _ = quadratic_descent(2, eta, 2, 8)
    ax.plot(range(len(x)), x, marker="o", label=f"eta={eta}: {descent_regime(eta,2)}")
ax.set(xlabel="Iteration", ylabel="x", title="Gradient descent: scalar quadratic")
ax.legend(fontsize=8); ax.grid(alpha=.2); plt.tight_layout(); plt.show()
```

## 4 / Linear regression
Compare the gradient descent estimate to the least-squares solution. / Comparer les deux méthodes.

```python
t = np.linspace(-1,1,5)
X = np.column_stack([np.ones_like(t), t]); y = 1 + 2*t
w, loss = linear_regression_descent(X, y, .5, 100)
print("Gradient descent:", w[-1]); print("Least squares:", least_squares(X,y))
np.testing.assert_allclose(w[-1], least_squares(X,y), atol=1e-8)
plt.plot(loss); plt.yscale("log"); plt.xlabel("Iteration"); plt.ylabel("Loss"); plt.title("Convergence on synthetic data"); plt.show()
```
